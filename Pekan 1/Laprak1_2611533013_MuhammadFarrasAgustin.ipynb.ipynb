## 2.1 Dasar Teori

Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub membahas dasar penggunaan bahasa pemrograman Python, Visual Studio Code sebagai code editor, serta GitHub sebagai media penyimpanan dan pengelolaan kode program.

1.Python

Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana sehingga mudah dipahami oleh pemula. Python dapat digunakan untuk berbagai kebutuhan, seperti pembuatan aplikasi, pengolahan data, dan pengembangan perangkat lunak.

2.Visual Studio Code

Visual Studio Code (VS Code) merupakan code editor yang digunakan untuk menulis, mengedit, dan menjalankan kode program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python, serta menyediakan fitur seperti terminal, debugging, dan integrasi dengan Git.

3.Git

Git merupakan sistem version control yang digunakan untuk mencatat perubahan pada kode program. Dengan Git, programmer dapat mengelola berbagai versi program dan mengetahui perubahan yang telah dilakukan.

4.GitHub

GitHub merupakan platform berbasis daring yang digunakan untuk menyimpan dan mengelola repository Git. GitHub dapat membantu programmer dalam melakukan pencadangan kode, mengakses proyek dari berbagai perangkat, serta bekerja sama dalam pengembangan program.


5.Repository

Repository merupakan tempat penyimpanan proyek yang berisi kode program dan berbagai berkas pendukung. Repository dapat disimpan secara lokal pada komputer maupun secara daring pada GitHub.

6.Commit dan Push

Commit digunakan untuk menyimpan perubahan yang telah dilakukan pada repository lokal. Push digunakan untuk mengirim perubahan dari repository lokal ke repository yang berada di GitHub.

### 2.2.2 Membuat Repository Github

Setelah program dibuat dan diuji pada VS Code, proyek dihubungkan ke platform GitHub. Repositori ini berfungsi untuk menyimpan, mengelola, dan mengamankan berkas kode program secara daring (online).

Tahapan yang dilakukan yaitu:

1.Membuat repositori baru pada halaman GitHub.

2.Membuka folder proyek menggunakan VS Code.

3.Membuka terminal terintegrasi pada VS Code.

4.Menginisialisasi Git pada folder proyek lokal.

5.Menambahkan berkas ke dalam staging area.

6.Membuat commit untuk menyimpan riwayat perubahan.

7.Menghubungkan repositori lokal dengan repositori di GitHub.

8.Melakukan push untuk mengunggah kode ke GitHub.

Contoh perintah (command) yang digunakan pada terminal:

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

1. git init: Digunakan untuk membuat dan menginisialisasi repositori Git baru pada folder proyek lokal.
2. git add .: Digunakan untuk menambahkan seluruh berkas dan perubahan pada proyek ke dalam staging area.
3. git commit: Digunakan untuk menyimpan riwayat perubahan dari staging area disertai pesan deskriptif.
4. git branch -M main: Digunakan untuk mengubah atau mengonfirmasi nama cabang (branch) utama menjadi main.
5. git remote add origin: Digunakan untuk menghubungkan repositori lokal dengan repositori yang telah dibuat di GitHub.
6. git push: Digunakan untuk mengunggah berkas dan riwayat commit dari komputer lokal ke reposi
