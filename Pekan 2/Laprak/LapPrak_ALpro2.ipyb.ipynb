## 2.1 Dasar Teori
1. Integer adalah tipe data yang digunakan untuk menyimpan bilangan bulat, yaitu bilangan tanpa angka desimal. Contohnya adalah 10, 25, -5, dan 100

2. Float adalah tipe data yang digunakan untuk menyimpan bilangan yang memiliki nilai desimal. Contohnya 3.14, 10.5, dan -2.75

3. Complex adalah tipe data yang digunakan untuk menyimpan bilangan kompleks, yaitu bilangan yang memiliki bagian real dan imajiner.contohnya adalah 3+4j

4. String adalah tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String ditulis menggunakan tanda petik tunggal (' ') atau tanda petik ganda (" ").contohnya "Selamat datang!" dan "Teknik Informatika"

5. Boolean adalah tipe data yang hanya memiliki dua nilai, yaitu True dan False. Boolean biasanya digunakan dalam kondisi atau percabangan untuk menentukan apakah suatu pernyataan benar atau salah.

6. Konstanta adalah nilai yang ditetapkan untuk tidak berubah selama program dijalankan. Dalam Python tidak terdapat aturan khusus yang benar-benar mengunci nilai menjadi konstanta. Biasanya konstanta ditulis menggunakan huruf kapital sebagai tanda bahwa nilainya sebaiknya tidak diubah.contohnya PI = 3.14

7. Variabel adalah tempat untuk menyimpan data atau nilai yang dapat digunakan dan diubah selama program berjalan. Variabel memiliki nama dan nilai yang diberikan kepadanya. Contohnya nama = "Farras" atau umur = 20

8. Output adalah proses menampilkan informasi atau hasil pengolahan data kepada pengguna. Dalam Python, output biasanya dilakukan menggunakan fungsi print().

9. Input adalah proses memasukkan data dari pengguna ke dalam program. Dalam Python, input dapat dilakukan menggunakan fungsi input().


## 2.2 Kode Program

### 2.2.1 Program Boolean_2611533013


```python
# Buat file dengan nama Boolean_NIM.py
# Nama variabel ditambah 4 digit terakhir NIM contoh: nilai_1234
# Deklarasi variable dengan tipe data Boolean
is_lulus = True
is_cumlaude = True

# Menggunakan Boolean
nilai_3013 = 85
batas_lulus_3013 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan = nilai_3013 >= batas_lulus_3013 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_3013)
print("Apakah Lulus:", status_kelulusan)
if is_lulus and is_cumlaude:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

**Penjelasan :**
1. Deklarasi is_lulus = True sebagai tipe data Boolean.

2. Deklarasi is_cumlaude = True sebagai tipe data Boolean.

3. Deklarasi nilai_3013 = 85 sebagai tipe data Integer.

4. Deklarasi batas_lulus_3013 = 75 sebagai tipe data Integer.

5. Menentukan variabel status_kelulusan dari ekspresi perbandingan nilai_3013 >= batas_lulus_3013 yang menghasilkan nilai Boolean (True).

6. Cetak ("=== Check Kelulusan ===").

7. Cetak ("Nilai: ", nilai_3013).

8. Cetak ("Apakah Lulus? :", status_kelulusan).

9. Pengujian kondisi if is_lulus and is_cumlaude: untuk mengecek dua variabel Boolean sekaligus.

10. Cetak ("Selamat, Anda lulus dengan predikat Cum laude!") jika kondisi pengujian bernilai True.

### 2.2.2 Program Input_2611533013

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk  menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
# program ini menggunakan fungsi input () untuk meminta pengguna
# masukkan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_3013 = input ('Masukkan nilai panjang: ')
lebar_3013 = input ('Masukkan nilai lebar: ')
luas_3013 = int(panjang_3013) * int(lebar_3013)
print("Luas =", luas_3013)
```

**Penjelasan :**

3. Melakukan konversi tipe data (type casting) panjang_3013 dan lebar_3013 menjadi int, mengalikan keduanya, lalu menyimpannya pada variabel luas_3013.

4. Cetak ("Luas =", Luas_3013).

### 2.2.3 Program Konstanta_2611533013.

```python
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3013 = float(input('Masukkan nilai jari jari:'))
luas_3013 = PI * jari_3013 * jari_3013
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3013, luas_3013))
```

**Penjelasan :**

1. Mengimpor entitas Final dari modul typing.

2. Deklarasi PI: Final = 3.14 sebagai variabel konstanta berjenis Float.

3. Cetak nilai PI dengan format bilangan desimal/float ("pi: %f" % (PI)).

4. Mengambil masukan nilai jari-jari dari pengguna dengan input(), dikonversi ke Float, lalu disimpan pada variabel jari_3013.

5. Menghitung luas lingkaran dengan rumus PI * jari_3013 * jari_3013 lalu menyimpannya pada variabel luas_3013.

6. Cetak hasil perhitungan luas lingkaran beserta jari-jarinya dengan format dua angka di belakang koma (%.2f).

### 2.2.4 Program Numerik_2611533013.

```python
# Buat file dengan nama Numerik_NIM.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
#create a variable with numeric type value
a_3013=100
print("The type of variable having value" , a_3013, "is " , type(a_3013))

#create a variable with float value
b_3013=10.2345
print("The type of variable having value" , b_3013 , " is " , type (b_3013))

#create a variable with complex value.
c_3013=100+3j
print("The type of variable having value" , c_3013 , " is " , type (c_3013))
```

**Penjelasan :**

1. Deklarasi a_3013 = 100 sebagai tipe data Integer.

2. Cetak nilai a_3013 beserta tipe datanya menggunakan fungsi type(a_3013).

3. Deklarasi b_3013 = 10.2345 sebagai tipe data Float.

4. Cetak nilai b_3013 beserta tipe datanya menggunakan fungsi type(b_3013).

5. Deklarasi c_3013 = 100+3j sebagai tipe data Complex.

6. Cetak nilai c_3013 beserta tipe datanya menggunakan fungsi type(c_3013).
